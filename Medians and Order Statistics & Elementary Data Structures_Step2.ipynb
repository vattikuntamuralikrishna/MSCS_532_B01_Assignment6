# Arrays and Matrices Implementation

We implemented a fixed-capacity CustomArray (handling element shifting for insertion/deletion) and a row-major 2D Matrix.

```python
from typing import Any, List


# 1. CUSTOM ARRAY (Fixed Capacity, Contiguous Memory)


class CustomArray:# Fixed-capacity array with explicit manual shifting for insertion and deletion.
    
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.size = 0
        self.data = [None] * capacity

    def access(self, index: int) -> Any: #Access: O(1) direct lookup by index.
        
        if not 0 <= index < self.size:
            raise IndexError("Array index out of bounds.")
        return self.data[index]

    def insert(self, index: int, value: Any) -> None: #Insertion: O(n) worst-case due to rightward element shifting.
        
        if self.size >= self.capacity:
            raise MemoryError("Array capacity reached.")
        if not 0 <= index <= self.size:
            raise IndexError("Array index out of bounds.")
        
        # Shifting elements to the right to create space at index
        for i in range(self.size, index, -1):
            self.data[i] = self.data[i - 1]
            
        self.data[index] = value
        self.size += 1

    def delete(self, index: int) -> Any: #Deletion: O(n) worst-case due to leftward element shifting.
        if not 0 <= index < self.size:
            raise IndexError("Array index out of bounds.")
        
        removed_value = self.data[index]
        
        # Shifting elements to the left to close the gap
        for i in range(index, self.size - 1):
            self.data[i] = self.data[i + 1]
            
        self.data[self.size - 1] = None
        self.size -= 1
        return removed_value


# 2. MATRIX (2D Grid Structure)

class Matrix: #2D Row-Major Matrix supporting Insertion (Write), Deletion (Reset), and Access.
    
    def __init__(self, rows: int, cols: int, default_val: Any = 0):
        self.rows = rows
        self.cols = cols
        self.data = [[default_val for _ in range(cols)] for _ in range(rows)]

    def access(self, row: int, col: int) -> Any: #Access: O(1) lookup at coordinate (row, col).
        
        if not (0 <= row < self.rows and 0 <= col < self.cols):
            raise IndexError("Matrix coordinates out of bounds.")
        return self.data[row][col]

    def insert(self, row: int, col: int, value: Any) -> None: #Insertion/Update: O(1) write to coordinate (row, col).
        if not (0 <= row < self.rows and 0 <= col < self.cols):
            raise IndexError("Matrix coordinates out of bounds.")
        self.data[row][col] = value

    def delete(self, row: int, col: int, default_val: Any = 0) -> Any: #Deletion: O(1) reset of cell value at coordinate (row, col)

        if not (0 <= row < self.rows and 0 <= col < self.cols):
            raise IndexError("Matrix coordinates out of bounds.")
        removed_val = self.data[row][col]
        self.data[row][col] = default_val
        return removed_val


```

```python
# Validation

if __name__ == "__main__":
    print("1. CUSTOM ARRAY DEMONSTRATION")
    
    arr = CustomArray(capacity=5)
    
    # --- INSERTION ---
    arr.insert(0, 10)  # [10]
    arr.insert(1, 30)  # [10, 30]
    arr.insert(1, 20)  # Insert 20 at index 1 -> [10, 20, 30]
    print("Array after insertions:")
    print("  Contents:", [arr.access(i) for i in range(arr.size)])
    
    # --- ACCESS ---
    val_at_1 = arr.access(1)
    print(f"\n Accessing element at index 1: {val_at_1}")
    
    # --- DELETION ---
    deleted_val = arr.delete(1)  # Remove index 1 (value 20)
    print(f"\nDeleted element at index 1 (value: {deleted_val})")
    print("Array after deletion:")
    print("  Contents:", [arr.access(i) for i in range(arr.size)])

    print("2. MATRIX DEMONSTRATION")
    
    mat = Matrix(rows=3, cols=3, default_val=0)
    
    # --- INSERTION ---
    mat.insert(0, 1, 42)  # Insert 42 at Row 0, Col 1
    mat.insert(2, 2, 99)  # Insert 99 at Row 2, Col 2
    print("Matrix state after insertions:")
    for r in range(mat.rows):
        print(f"  Row {r}: {[mat.access(r, c) for c in range(mat.cols)]}")
        
    # --- ACCESS ---
    val = mat.access(0, 1)
    print(f"\nAccessing element at (Row 0, Col 1): {val}")
    
    # --- DELETION ---
    deleted_mat_val = mat.delete(0, 1)
    print(f"\nDeleted element at (Row 0, Col 1) (reset value back from {deleted_mat_val} to 0)")
    print("Matrix state after deletion:")
    for r in range(mat.rows):
        print(f"  Row {r}: {[mat.access(r, c) for c in range(mat.cols)]}")
```

# Stacks and Queues: Implementation

For the ArrayQueue, we implement a Circular Array using front and rear pointers so that enqueue and dequeue operations run in O(1) time without requiring O(n) element shifting.

```python
from typing import Any, List

# 1. ARRAY-BASED STACK (LIFO: Last-In, First-Out)

class ArrayStack: #Stack implementation using a dynamic array base.
    
    def __init__(self):
        self.items: List[Any] = []

    def push(self, element: Any) -> None: #Insertion: Push element onto the top of the stack - O(1) amortized.
        self.items.append(element)

    def pop(self) -> Any: #Deletion: Remove and return the top element - O(1).
        if self.is_empty():
            raise IndexError("Stack Underflow: Cannot pop from an empty stack.")
        return self.items.pop()

    def peek(self) -> Any: #Access: View the top element without removing it - O(1).
        if self.is_empty():
            raise IndexError("Stack Empty: Cannot peek.")
        return self.items[-1]

    def is_empty(self) -> bool: #Check if the stack has no elements.
        return len(self.items) == 0

    def size(self) -> int: #Return total number of elements in the stack.
        return len(self.items)


# 2. CIRCULAR ARRAY-BASED QUEUE (FIFO: First-In, First-Out)

class CircularArrayQueue: #Queue implementation using a fixed-capacity circular array to achieve O(1) operations.
    
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.data = [None] * capacity
        self.front = 0
        self.rear = 0
        self.count = 0

    def enqueue(self, element: Any) -> None: #Insertion: Add element to the rear of the queue - O(1).
        if self.count == self.capacity:
            raise MemoryError("Queue Overflow: Capacity reached.")
        self.data[self.rear] = element
        self.rear = (self.rear + 1) % self.capacity
        self.count += 1

    def dequeue(self) -> Any: #Deletion: Remove and return the front element - O(1).
        
        if self.is_empty():
            raise IndexError("Queue Underflow: Cannot dequeue from an empty queue.")
        removed_val = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.count -= 1
        return removed_val

    def peek(self) -> Any: #Access: View the front element without removing it - O(1).
        if self.is_empty():
            raise IndexError("Queue Empty: Cannot peek.")
        return self.data[self.front]

    def is_empty(self) -> bool: #Check if the queue has no elements.
        return self.count == 0

    def size(self) -> int: #Return total number of elements in the queue.
        return self.count

```

```python
# Validation

if __name__ == "__main__":
    
    print("1. ARRAY STACK DEMONSTRATION (LIFO)")
    
    stack = ArrayStack()
    
    # --- INSERTION (PUSH) ---
    print("Pushing elements: 10, 20, 30 onto stack...")
    stack.push(10)
    stack.push(20)
    stack.push(30)
    print(f"Current Stack Size: {stack.size()}")
    
    # --- ACCESS (PEEK) ---
    print(f"Accessing top element (peek): {stack.peek()}")
    
    # --- DELETION (POP) ---
    popped = stack.pop()
    print(f"Popped top element (deletion): {popped}")
    print(f"New top element after pop: {stack.peek()}")
    print(f"Stack size after pop: {stack.size()}")

    print("\n")
    print("2. CIRCULAR ARRAY QUEUE DEMONSTRATION (FIFO)")
    
    queue = CircularArrayQueue(capacity=4)
    
    # --- INSERTION (ENQUEUE) ---
    print("Enqueueing elements: 'Task A', 'Task B', 'Task C' into queue...")
    queue.enqueue("Task A")
    queue.enqueue("Task B")
    queue.enqueue("Task C")
    print(f"Current Queue Size: {queue.size()}")
    
    # --- ACCESS (PEEK) ---
    print(f"Accessing front element (peek): '{queue.peek()}'")
    
    # --- DELETION (DEQUEUE) ---
    dequeued = queue.dequeue()
    print(f"Dequeued front element (deletion): '{dequeued}'")
    print(f"New front element after dequeue: '{queue.peek()}'")
    print(f"Queue size after dequeue: {queue.size()}")
    
    # --- DEMONSTRATING CIRCULAR WRAPAROUND ---
    print("\nEnqueueing 'Task D' and 'Task E' to test circular buffer wrap...")
    queue.enqueue("Task D")
    queue.enqueue("Task E")
    print(f"Current Front element: '{queue.peek()}'")
    print(f"Total elements queued: {queue.size()}/{queue.capacity}")
```

# Linked list 

A SinglyLinkedList composed of explicit Node pointers with support for head/tail insertions, value deletion, and list traversal.

```python
from typing import Any, List, Optional

# 1. NODE BUILDING BLOCK

class Node: #Individual node holding data and a pointer to the next node.
    
    def __init__(self, value: Any):
        self.value = value
        self.next: Optional['Node'] = None



# 2. SINGLY LINKED LIST DATA STRUCTURE

class SinglyLinkedList: #Singly Linked List with explicit pointer-manipulation methods.
    
    def __init__(self):
        self.head: Optional[Node] = None

    def insert_head(self, value: Any) -> None: #Insertion at Head: O(1) time complexity.
        new_node = Node(value)
        new_node.next = self.head
        self.head = new_node

    def insert_tail(self, value: Any) -> None: #Insertion at Tail: O(n) time complexity (requires traversing to the end).
        new_node = Node(value)
        if not self.head:
            self.head = new_node
            return
        
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = new_node

    def delete_value(self, value: Any) -> bool: #Deletion by Value: O(n) search time, O(1) pointer updates upon location.
        if not self.head:
            return False
        
        # Case 1: Target node is the head node
        if self.head.value == value:
            self.head = self.head.next
            return True

        # Case 2: Target node is downstream
        curr = self.head
        while curr.next and curr.next.value != value:
            curr = curr.next

        if curr.next:  # Match found
            curr.next = curr.next.next
            return True
            
        return False  # Target value not found in list

    def traverse(self) -> List[Any]: #Traversal: Visits every node from head to tail - O(n).
        elements = []
        curr = self.head
        while curr:
            elements.append(curr.value)
            curr = curr.next
        return elements

    def search(self, value: Any) -> bool: #Access/Search: Determines if a value exists - O(n).
        curr = self.head
        while curr:
            if curr.value == value:
                return True
            curr = curr.next
        return False

```

```python
# Validation

if __name__ == "__main__":

    print("SINGLY LINKED LIST DEMONSTRATION")
    
    sll = SinglyLinkedList()

    # --- INSERTION ---
    print("1. INSERTION OPERATIONS:")
    print("   -> Inserting 20 at head")
    sll.insert_head(20)
    print("   -> Inserting 10 at head")
    sll.insert_head(10)
    print("   -> Inserting 30 at tail")
    sll.insert_tail(30)
    print("   -> Inserting 40 at tail")
    sll.insert_tail(40)

    # --- TRAVERSAL ---
    print("\n2. TRAVERSAL OPERATION:")
    print("   Current List Sequence:", sll.traverse())

    # --- SEARCH / ACCESS ---
    print("\n3. SEARCH / ACCESS OPERATIONS:")
    print("   Is 30 in the list?", sll.search(30))
    print("   Is 99 in the list?", sll.search(99))

    # --- DELETION ---
    print("\n4. DELETION OPERATIONS:")
    print("   -> Deleting middle element (20)...")
    sll.delete_value(20)
    print("      Updated List:", sll.traverse())
    
    print("   -> Deleting head element (10)...")
    sll.delete_value(10)
    print("      Updated List:", sll.traverse())
    
    print("   -> Attempting to delete non-existent element (99)...")
    found = sll.delete_value(99)
    print(f"      Deletion status: {found}")
    print("      Final List State:", sll.traverse())
```

# Rooted Trees

A general N-ary Rooted Tree implementation using nodes linked to lists of child node pointers, along with a Depth-First Search (DFS) traversal.

```python
from typing import Any, List, Optional
from collections import deque


# 1. TREE NODE (Linked Children Structure)

class TreeNode: #Node structure for general rooted trees using linked child references.
    
    def __init__(self, value: Any):
        self.value = value
        self.children: List['TreeNode'] = []

    def add_child(self, child_node: 'TreeNode') -> None: #Insertion: Adds a child reference link to this node - O(1).
        self.children.append(child_node)

    def remove_child(self, target_value: Any) -> bool: #Deletion: Removes a child node and its entire subtree - O(c) where c is child count.
        for idx, child in enumerate(self.children):
            if child.value == target_value:
                del self.children[idx]
                return True
        return False


# 2. ROOTED TREE WRAPPER

class RootedTree: #General Rooted Tree operations wrapper.
    
    def __init__(self, root_val: Any):
        self.root = TreeNode(root_val)

    def insert(self, parent_val: Any, new_val: Any) -> bool: #Insertion: Locates target parent node and appends a new child node.
        parent_node = self.find(self.root, parent_val)
        if parent_node:
            parent_node.add_child(TreeNode(new_val))
            return True
        return False

    def delete(self, value: Any) -> bool: #Deletion: Prunes target subtree starting at matching value.
        if self.root.value == value:
            raise ValueError("Cannot delete the root node via prune operation.")
        return self._delete_helper(self.root, value)

    def _delete_helper(self, current: TreeNode, value: Any) -> bool:
        if current.remove_child(value):
            return True
        for child in current.children:
            if self._delete_helper(child, value):
                return True
        return False

    def find(self, current: Optional[TreeNode], value: Any) -> Optional[TreeNode]: #Access / Search: Recursively locates a node matching value - O(n).
        if not current:
            return None
        if current.value == value:
            return current
        
        for child in current.children:
            res = self.find(child, value)
            if res:
                return res
        return None

    def traverse_dfs(self, current: Optional[TreeNode] = None) -> List[Any]: #Traversal (DFS): Pre-order recursive depth-first exploration - O(n).
        if current is None:
            current = self.root
        
        result = [current.value]
        for child in current.children:
            result.extend(self.traverse_dfs(child))
        return result

    def traverse_bfs(self) -> List[Any]: #Traversal (BFS): Level-order queue-based breadth-first exploration - O(n).
        result = []
        if not self.root:
            return result
        
        queue = deque([self.root])
        while queue:
            curr = queue.popleft()
            result.append(curr.value)
            for child in curr.children:
                queue.append(child)
        return result

```

```python
# Validation

if __name__ == "__main__":
    
    print("ROOTED TREE DEMONSTRATION")
    
    # --- CREATION & INSERTION ---
    print("1. CREATING TREE & INSERTING NODES...")
    tree = RootedTree("Root")
    

    tree.insert("Root", "FolderA")
    tree.insert("Root", "FolderB")
    tree.insert("FolderA", "FileA1")
    tree.insert("FolderA", "FileA2")
    
    print("   Inserted: 'FolderA' & 'FolderB' under 'Root'")
    print("   Inserted: 'FileA1' & 'FileA2' under 'FolderA'")

    # --- TRAVERSAL ---
    print("\n2. TREE TRAVERSALS:")
    print("   Depth-First Search (DFS - Pre-order):", tree.traverse_dfs())
    print("   Breadth-First Search (BFS - Level-order):", tree.traverse_bfs())

    # --- ACCESS / SEARCH ---
    print("\n3. SEARCH / ACCESS OPERATIONS:")
    node_found = tree.find(tree.root, "FileA1")
    print(f"   Search for 'FileA1': Found node with value '{node_found.value if node_found else None}'")
    node_missing = tree.find(tree.root, "FileZ")
    print(f"   Search for non-existent 'FileZ': Found -> {node_missing}")

    # --- DELETION / PRUNING ---
    print("\n4. DELETION (SUBTREE PRUNING):")
    print("   -> Deleting node 'FileA2'...")
    tree.delete("FileA2")
    print("      DFS Traversal after deleting 'FileA2':", tree.traverse_dfs())
    
    print("   -> Deleting entire subtree 'FolderA'...")
    tree.delete("FolderA")
    print("      Final DFS Traversal (FolderA and its contents removed):", tree.traverse_dfs())
```
